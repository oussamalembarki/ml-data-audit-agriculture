```python
import pandas as pd

# Charger le dataset
df = pd.read_csv('Crop_recommendation.csv')

# Afficher les 5 premières lignes
display(df.head())
```

```python
# 1. Vérifier s'il y a des valeurs manquantes (NaN / Null)
print("--- Valeurs manquantes par colonne ---")
print(df.isnull().sum())

# 2. Vérifier s'il y a des lignes dupliquées
print(f"\n--- Nombre de lignes en double : {df.duplicated().sum()} ---")

# 3. Afficher les types de données (pour vérifier s'il y a du texte là où il faut des chiffres)
print("\n--- Types des colonnes ---")
print(df.dtypes)

# 4. Statistiques descriptives (pour repérer d'éventuelles valeurs aberrantes / min-max bizarres)
print("\n--- Statistiques descriptives ---")
display(df.describe())
```
